```python
# find the  total sales per year      

import numpy as np

sales_data=np.array([
     [1,150000,180000,220000,250000], #karim (id,2021,2022,2023,2024)
     [2,120000,140000,160000,190000], # dominos (id,2021,2022,2023,2024)
     [3,200000,230000,260000,300000], # kfc  (id,2021,2022,2023,2024)
     [4,180000,210000,240000,270000],  # aslma chiken  (id,2021,2022,2023,2024)
     [5,160000,185000,205000,230000]   # javed nahari   (id,2021,2022,2023,2024)
])

# first of all we to do slicing 

required_array=sales_data[:,1:]
# print(required_array)
total_sales_per_year=np.sum(required_array,axis=0)
print(total_sales_per_year)
```

```python
# quest : calculate the  total sales per restaurent 

import numpy as np

sales_data=np.array([
     [1,150000,180000,220000,250000], #karim
     [2,120000,140000,160000,190000],
     [3,200000,230000,260000,300000],
     [4,180000,210000,240000,270000],
     [5,160000,185000,205000,230000]
])

# slicing 

required_array=sales_data[0:,1:]

total_sales_per_res=np.sum(required_array,axis=1)
print(total_sales_per_res)


```

```python
# ques :calculate the min and max sales per restaurent 
import numpy as np

sales_data=np.array([
     [1,150000,180000,220000,250000],  #150000
     [2,120000,140000,160000,190000],  # 120000
     [3,200000,230000,260000,300000],  #200000
     [4,180000,210000,240000,270000],  #150000
     [5,160000,185000,205000,230000]   #160000

])
# slicing 
required_array=sales_data[:,1:]
min_sales_per_res=np.min(required_array,axis=1) #[15k,120k,2lakh,150k,160k]
print(min_sales_per_res)
```

```python
# avg sales per restaurent 

import numpy as np

sales_data=np.array([
     [1,150000,180000,220000,250000],
     [2,120000,140000,160000,190000],
     [3,200000,230000,260000,300000],
     [4,180000,210000,240000,270000],
     [5,160000,185000,205000,230000]
])

# slicing 

required_array=sales_data[:,1:]
avg_sales_per_restaurent=np.mean(required_array,axis=1)
print(avg_sales_per_restaurent) #[20000,152,500]
```
