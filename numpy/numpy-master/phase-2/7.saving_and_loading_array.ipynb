## save the numpy array in a file 

```python
import numpy as np

arr1=np.array([1,2,5,10])

arr2=np.arange(2,10,1)

arr3=np.zeros((2,3)) # it will create  a numpy array with values zeroes but for diemtnion or shape you have to mention it 

arr4=np.ones(2,4) # it will create a numpy array with values ones but for dimention or shape you have to mention it 

np.save("array1.npy",arr1)
np.save("array2.npy",arr2)
np.save("array3.npy",arr3)

np.save("array4.npy",arr4)
```

## load the saved numpy array 

```python
loadded_array1=np.load("array1.npy")
print(loadded_array1)
```
