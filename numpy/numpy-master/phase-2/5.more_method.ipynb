## flatten(),reshape,ravel and transpose method 

```python
import numpy as np
arr=np.arange(1,19,1) # it will create a one dimentinal array with sequence of number

print(arr)
print(arr.shape) # (18,)


# reshape :that means can change the shape of current numpy array 

re_shaped=arr.reshape(3,3,2)  
print("reshaped to (3,3,2)",re_shaped)


# flatten() :convert multi dimentinal array to one dimentinal array

flatten=re_shaped.flatten()
print("flatten the reshaped array:",flatten)
```

## flatten()   vs ravel()




```python
import numpy as np 
num_arr=np.arange(2,20,1)  # return one deimentinal numpy array
print("original array",num_arr)

# reshape 

reshapped=num_arr.reshape(3,3,2)
print("reshapped(3,3,2):",reshapped)

# flatten

flatten=reshapped.flatten() # it will return the copy of original array after flattern
print("flatten:",flatten)

# ravel

ravel=reshapped.ravel()   # it will return the orignal array after the flatternn 
print("ravel:",ravel)



```

## transpose 
## def: it will convert row to column or column to row 

```python
import numpy as np

num_2darr=np.array([
  [10,20,40,60],
  [50,60,40,60],
  [70,20,90,60],
])

print("before transpose:",num_2darr)

transpose_arr=np.transpose(num_2darr)

print("transpose_arr:",transpose_arr)
```
