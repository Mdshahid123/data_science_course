## adding two numpy array

```python
import numpy as np

num_arr1=np.array([10,40,50])
num_arr2=np.array([12,11,8])

print(num_arr1 + num_arr2 ) # it will add the corresponding element 

combined_arr=np.concatenate((num_arr1,num_arr2)) # it join  the array
print(combined_arr)
```

## deleting 

```python
import numpy as np

num_arr=np.array([10,40,50])
arr_after_deletion=np.delete(num_arr,1) # it will return the remaing array after deletion
print(arr_after_deletion)

```

## checking compatiblity

## checking compatiblity means comaring a shape of an array

```python
import numpy as np


# creating a one dimention numpy array

num_arr1=np.array([10,40,50])
num_arr2=np.array([12,11,8])

print("shape of arr1:",num_arr1.shape) #(3)
print("shape of arr2:",num_arr2.shape) #(3)

print(num_arr1.shape==num_arr2.shape) #True


```

```python
import numpy as np


# creating  a two dimention numpy array

num_2darr1=np.array([
   [10,20,40,60],
   [50,70,80,90],
   [0,10,30,40]
])
num_2darr2=np.array([
   [80,70,60,50],
   [80,70,60,5],
   [80,71,60,50],
])

print("shape of arr1:",num_2darr1.shape) #(3,4)
print("shape of arr2:",num_2darr2.shape) #(3,4)

print(num_2darr1.shape==num_2darr2.shape) #True
```
