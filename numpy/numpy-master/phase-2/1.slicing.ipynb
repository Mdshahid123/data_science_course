##  SLICING IN NUMPY ARRAY

#### In NumPy, slicing means extracting a part of an array.

#### The basic syntax is:

#### array[start idx  : stop index : step]
#### start inx → from where to start
#### stop idx → where to stop (not included)
#### step → how many positions to jump


### 1) Slicing in one-D numpy array

```python
# slicing a one-D numpy array

import numpy as np
num_arr=np.array([1,2,3,4,10,30,50])

print(f"numpy-array:{num_arr}")


# let's do slicing  with step-1

s1=num_arr[1:4:1] 
print(f" i want slice 2,3,4 : {s1}")

s2=num_arr[3:6]
print(f"i want slice 4,10,30:{s2}")


# do slicing with step 2
s3=num_arr[1:7:2] 
print(f"with two step:{s3}")



# similarly we can get any slice with any step from the given numpy array
```

### 2) Slicing a 2D NumPy Array

### syntax

### arr[row_start:row_stop, column_start:column_stop]

```python
import numpy as np 
num_arr=np.array([

  [10,20,30,40]
  ,
  [40,50,70,80]
  ,
  [3,4,5,56]
  ,

])

print(num_arr[0:3,1:3])

```
