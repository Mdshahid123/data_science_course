##  NumPy, filtering means selecting elements from an array based on a condition.  

## syntaxt 

## array_name[condition] ,where condtion should be expresstion

##    or 

## np.where(condtion)


```python
# select number greater than 10

import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = arr[arr > 10]

print(result)



```

```python
# values eqal to 10

import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = arr[arr == 10]

print(result)



```

```python
import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = arr[(arr > 5) & (arr < 20)]

print(result)
```

```python
import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = arr[(arr > 5) & (arr < 20)]

print(result)
```

```python
# even number 
import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = arr[arr % 2==0]

print(result)
```

```python
# odd number 
import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = arr[arr % 2!=0]

print(result)
```

## using np.where()

```python
import numpy as np

arr = np.array([10, 5, 20, 3, 15])

result = np.where(arr%2==0) # it will return the indexes

print(result)
print(arr[result])
```
