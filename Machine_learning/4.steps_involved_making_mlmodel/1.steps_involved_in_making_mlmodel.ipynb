## What is EDA?
``` 

-EDA (Exploratory Data Analysis) is the process of exploring and understanding a dataset in order to identify patterns, relationships, and potential problems in the data.


-During EDA, we may find patterns such as:

    Increasing relationship → when one variable increases, another variable also tends to increase.
    Decreasing relationship → when one variable increases, another variable tends to decrease.
    Distribution patterns → how the values are spread across the dataset.
    Group/category patterns → how data is distributed among different categories.

-- Problems in Data

During EDA, we may also identify problems such as:

   Missing values
   Duplicate values
   Incorrect or inconsistent values
   Outliers
   Incorrect data types
   Invalid data

-- In simple words:

EDA = Explore the data → Understand the data → Find patterns/relationships → Identify problems → Get insights


```



## steps invloved in EDA 


```

```

option1: fill the missing values with mean/median if column is numerical


option2:  fill the missing values with mode if column is categorical

option3:  use algorithm like liner regression ,KNN,interpolation 

option4:  remove the entire rows or columns containg  more than 50 % missing values if data is not important 


note:don't blindly drop all missing values. We first decide whether the missing information is important and how much data would be lost.

 ## 2) Remove Duplicate Rows

```
