## ReinForcement learning 

reienforcement learning is like trainig a dog or child  . you don't teach evry thing directly instead  you reward good behavior and punish bad behavour  and in this way child or dog learn from this exerinece and make better decision next time

Child does something correct → Reward 👍
Child does something wrong → Penalty 👎
Child learns from this experience → better decision next time

In ML terminology:

Agent → takes Action → Environment gives Reward/Penalty → Agent learns
