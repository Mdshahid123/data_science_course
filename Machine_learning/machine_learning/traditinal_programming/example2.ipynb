# Example2: Spam Email Detection

```
   Email	                        Label
 "You won $1000"	                Spam
 "Free prize for you"	            Spam
 "Click here to win"	            Spam
 "Meeting at 5 PM"	               Not Spam
"Please send the report"	       Not Spam

```



````python
def check_spam(email):
    spam_words = ["win money", "free prize", "click here"]

    for word in spam_words:
        if word in email.lower():
            return "Spam"

    return "Not Spam"

check_spam("free ")
    
````


```
Email(input data)
      ↓
   Program(rules)
      ↓
Spam / Not Spam(output)

```
