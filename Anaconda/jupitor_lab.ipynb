## what is jupytor lab 

#### jupytor lab is a web-based interactive development environment where we can write the  code, organize files, and view data visualizations in a flexible, all-in-one workspace


#### it comes with anaconda  distributer


```
JupyterLab
│
├── 📓 Notebook (.ipynb)
│      └── Write & run Python cell-by-cell
│
├── 📄 Python file (.py)
│      └── Write normal Python scripts
|
├── 📄 text file (.txt)
│      └── Write normal Python scripts
│      
│
├── 💻 Terminal
│      └── Run commands
|
|--- 💻 console
│      └── 
│
└── 📁 File Browser
       └── Manage your project files


``` 


## How launch the jupyter lab in anaconda distibuter 

### method-01 using anaconda prompt 

### -- open anaconda prompt and navigate to the project directory with which you want to work

#### -- and write the command "jupyter lab" it will be open the jupyter lab in browser


### method-02 using anaconda navigator

### -- serach anaconda navigator in window serbar and open the it 

### -- it will oepn the interface from where you hac launch it
